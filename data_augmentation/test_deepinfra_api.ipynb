```python
from openai import OpenAI
import os
from dotenv import load_dotenv
load_dotenv()  # loads variables from file .env

api_key = os.getenv("OPENAI_API_KEY")
api_key
```

```python
client = OpenAI(
    api_key=api_key,
    base_url="https://api.deepinfra.com/v1/openai",
)

def generate_from_prompt(prompt, model, system_msg="You are a helpful assistant."):
    completion = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": system_msg},
            {"role": "user", "content": prompt}
        ],
        max_tokens=100
    )

    return completion.choices[0].message.content

# Example
response = generate_from_prompt(
    prompt="Can you generate a biomedical text paragraph mentioning a genomic variant?",
    model="deepseek-ai/DeepSeek-V4-Flash"
)

print(response)
```
