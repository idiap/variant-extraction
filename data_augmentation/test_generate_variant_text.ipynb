```python
from openai import OpenAI
import os
from dotenv import load_dotenv
load_dotenv()  # loads variables from file .env

api_key = os.getenv("OPENAI_API_KEY")

client = OpenAI(
    api_key=api_key,
    base_url="https://api.deepinfra.com/v1/openai",
)
client
```

```python
import pandas as pd

pd.set_option("display.max_columns", None)

clinvar_df = pd.read_csv(
    "../resources/variant_summary.txt.gz",
    sep="\t",
    compression="gzip",
)
clinvar_df
```

```python
sampled_df = clinvar_df.sample(frac=0.0005, random_state=42).reset_index(drop=True)
sampled_df
```

```python
def generate_from_prompt(prompt, model, system_msg="You are a helpful assistant."):
    completion = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": system_msg},
            {"role": "user", "content": prompt}
        ],
        max_tokens=300
    )

    return completion.choices[0].message.content
```

```python
def generate_biomedical_text(variant, model):
    prompt = f"""
    Generate a biomedical text paragraph mentioning this genomic variant: {variant}
    """

    # Example
    response = generate_from_prompt(
        prompt=prompt,
        model=model
    )
    
    out = {
        "messages":
        [{
            "role": "system", 
            "content": "Extract only specific genetic variants from biomedical title and abstract text. Return only lines in the required format. Allowed outputs are:\nVariant: <mutation>, Gene: <gene>\nor exactly:\nNo variant\nDo not include explanations, markdown, JSON, or extra text."
        }, 
        {
            "role": "user",
            "content": response
        }, 
        {
            "role": "assistant",
            "content": f"Variant: {variant}"
        }]
    }
    return out

model = "XiaomiMiMo/MiMo-V2.5"

for i, variant in enumerate(clinvar_df['Name']):
    print("\n", i)
    out = generate_biomedical_text(variant, model)
    print(out["messages"][1]["content"])
    if i == 10:
        break
```
