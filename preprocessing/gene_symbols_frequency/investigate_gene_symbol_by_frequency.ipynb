```python
import pandas as pd

df = pd.read_csv(
    "../../resources/variant_summary.txt.gz",
    sep="\t",
    compression="gzip",
)
```

```python
gene_symbols = list(df['GeneSymbol'])
gene_symbols
```

```python
from collections import Counter

freq = Counter(gene_symbols)

freq
```

```python
import matplotlib.pyplot as plt

sorted_counts = [count for _, count in freq.most_common()]

plt.bar(range(1, len(sorted_counts) + 1), sorted_counts)
plt.xlabel("Rank")
plt.ylabel("Frequency")
plt.title("Frequency Distribution")
plt.show()
```

```python
print(sorted_counts[:10])  # top 10 frequencies
```

```python
threshold = 1000

filtered = {item: count for item, count in freq.items() if count > threshold}

len(filtered)
```
