```python
from collections import Counter

from transformers import AutoTokenizer


# ============================================================
# Configuration
# ============================================================

#MODEL_NAME = "meta-llama/Llama-3.1-8B-Instruct"  # change this
MODEL_NAME = "Qwen/Qwen2.5-7B-Instruct"

TOKENS_TO_CHECK = [
    # HGVS coordinate prefixes
    "c.", "p.", "g.", "m.", "n.", "r.",

    # HGVS operations / structural atoms
    "delins", "del", "dup", "ins", "inv", "con", "fs", "ext", "Ter",

    # rsID lexical prefix
    "rs",

    # Chromosome lexical prefix often seen in genomic coordinates
    "chr",

    # Three-letter amino-acid symbols
    "Ala", "Arg", "Asn", "Asp", "Cys", "Gln", "Glu", "Gly", "His", "Ile",
    "Leu", "Lys", "Met", "Phe", "Pro", "Ser", "Thr", "Trp", "Tyr", "Val",

    # A few common compact protein-change suffixes / frameshift endings
    "*", "=", "_", "+", "-",
]


# ============================================================
# Load tokenizer
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

vocab = tokenizer.get_vocab()


# ============================================================
# Check tokens
# ============================================================

num_tokens_counts = []

for token in TOKENS_TO_CHECK:

    in_vocab = token in vocab

    tokenized = tokenizer.tokenize(
        token,
        add_special_tokens=False,
    )

    token_ids = tokenizer.encode(
        token,
        add_special_tokens=False,
    )

    num_tokens = len(token_ids)

    # Store number of tokens for frequency calculation
    num_tokens_counts.append(num_tokens)

    '''print(f"Token: {token!r}")
    print(f"  Exact token in vocab: {in_vocab}")
    print(f"  Tokenization: {tokenized}")
    print(f"  Token IDs: {token_ids}")
    print(f"  Number of tokens: {num_tokens}")
    print()'''


# ============================================================
# Print number-of-tokens frequency distribution
# ============================================================

frequencies = Counter(num_tokens_counts)

print("=" * 50)
print("NUMBER OF TOKENS FREQUENCIES")
print("=" * 50)

for num_tokens, count in sorted(frequencies.items()):
    print(f"{num_tokens} token(s): {count}")


# ============================================================
# Optional: print percentages
# ============================================================

print()
print("=" * 50)
print("NUMBER OF TOKENS FREQUENCIES (%)")
print("=" * 50)

total = len(TOKENS_TO_CHECK)

for num_tokens, count in sorted(frequencies.items()):
    percentage = 100 * count / total

    print(
        f"{num_tokens} token(s): "
        f"{count} / {total} "
        f"({percentage:.2f}%)"
    )
```
